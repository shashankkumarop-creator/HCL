# 1. Introduction

## 1.1 Purpose
This System Requirements Specification (SRS) defines the functional, data, operational, and predictive model specifications for the E-Learning Engagement Analytics & Predictive Modeling Engine. The engine processes learner activity data from `18_elearning_engagement.csv` to evaluate course completion factors, calculate engagement indices, and cluster learner behaviors.

## 1.2 Scope
The software system automates data ingestion, cleaning, feature transformation, machine learning classification (predicting binary course completion), and unsupervised learner segmentation (K-Means clustering).

## 1.3 Acronyms & Definitions
- **EDA:** Exploratory Data Analysis.
- **Engagement Score:** A composite 0–100 index weighted across video consumption (40%), quiz score (40%), and forum activity (20%).
- **Watch Ratio:** Ratio of total hours watched to total course length hours (`Hours_Watched / Course_Length_Hours`).


# 2. Overall System Description

## 2.1 Architecture Overview
The system functions as an analytical processing pipeline operating over tabular enrollment records. Input dataset records pass through automated validation, missing-value imputation, numerical scaling, categorical encoding, baseline model training, and cluster analysis.

## 2.2 User Classes
- **Data Analysts:** Require repeatable workflows for exploratory data analysis, metric summaries, and correlation assessments.
- **Course Designers / Instructors:** Utilize predictive feature rankings and learner clusters to optimize course difficulty and content pacing.
- **Platform Managers:** Monitor overall completion rates across device types, pricing tiers, and subject areas.


# 3. Functional Requirements

## 3.1 Data Ingestion & Preprocessing Module

### REQ-DAT-01 — Schema Parsing
The system shall parse CSV files containing 12 schema columns:

`Enrollment_ID, Learner_ID, Course_Name, Course_Length_Hours, Hours_Watched, Quiz_Score, Quiz_Attempts, Forum_Posts, Course_Difficulty, Device_Type, Paid_Tier, Completed`

### REQ-DAT-02 — Deduplication
The pipeline shall automatically detect and purge exact duplicate enrollment records prior to feature computation (filtering 812 raw entries down to 800 unique enrollments).

### REQ-DAT-03 — Missing Value Imputation
Missing numerical values in `Hours_Watched` and `Quiz_Score` shall be imputed using non-parametric median values of their respective columns.


## 3.2 Feature Engineering & Scoring Engine

### REQ-FE-01 — Completion Target Encoding
The categorical target `Completed` ("Yes"/"No") shall be mapped to binary integers (1 and 0).

### REQ-FE-02 — Watch Ratio Computation
The system shall compute:

`Watch_Ratio = Hours_Watched / Course_Length_Hours`

for each learner record.

### REQ-FE-03 — Composite Engagement Index
The system shall construct a normalized 0–100 scale index defined by:

**Engagement_Score** = (`Hours_Watched` / max(`Hours_Watched`)) × 40 + (`Quiz_Score` / 100) × 40 + (`Forum_Posts` / max(`Forum_Posts`))) × 20


## 3.3 Predictive & Clustering Modules

### REQ-ML-01 — Classification Suite
The system shall train and evaluate Logistic Regression, Random Forest, and Gradient Boosting algorithms on a 80/20 stratified train/test split.

### REQ-ML-02 — Feature Importance Extraction
The system shall extract relative Gini feature importance rankings from Random Forest classifiers to identify key drivers of course completion.

### REQ-ML-03 — Learner Segmentation
The system shall execute K-Means clustering (k=3) on standardized engagement features (`Hours_Watched`, `Quiz_Score`, `Forum_Posts`, `Quiz_Attempts`) to partition learners into actionable behavioral segments.

# 5. Non-Functional Requirements

### NFR-PERF-01 — Execution Speed
Full pipeline execution (cleaning, model fitting, visualization rendering) shall complete within 15 seconds in standard Python cloud runtimes.

### NFR-ROB-01 — Input Fault Tolerance
Pipeline execution must safely handle missing entries without throwing unhandled exceptions or introducing data leaks between train/test splits.

### NFR-REP-01 — Reproducibility
All randomized ML routines (train/test splitting, Random Forest initialization, K-Means seed selection) must enforce explicit random state parameters (`random_state=42`).
